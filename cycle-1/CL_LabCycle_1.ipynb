# Question 1

Write regular expressions for the following languages and implement them using Python library. By “word”, we mean an alphabetic string separated from other words by whitespace, any relevant punctuation, line breaks, and so forth.

a. The set of all strings with two consecutive repeated words (e.g., “Humbert Humbert” and “the the” but not “the bug” or “the big bug”);

b. all strings that start at the beginning of the line with an integer and that end at the end of the line with a word;

c. all strings that have both the word grotto and the word raven in them (but not, e.g., words like grottos that merely contain the word grotto);

d. Write a pattern that places the first word of an English sentence in a
register. Deal with punctuation. The program reads input from a file and prints output to the terminal.

```python
import re

# Read input from file
filename = "input.txt"

with open(filename, "r", encoding="utf-8") as file:
    lines = file.readlines()

print("=" * 60)

# a. Strings with two consecutive repeated words
print("a. Consecutive Repeated Words:")
pattern_a = re.compile(r'\b(\w+)\s+\1\b', re.IGNORECASE)

for line in lines:
    match = pattern_a.search(line)
    if match:
        print(f"Matched: {line.strip()}")

print("=" * 60)

# b. Strings that start with an integer and end with a word
print("b. Starts with an Integer and Ends with a Word:")
pattern_b = re.compile(r'^\d+.*\b[A-Za-z]+\b\s*$', re.IGNORECASE)

for line in lines:
    if pattern_b.search(line.strip()):
        print(f"Matched: {line.strip()}")

print("=" * 60)

# c. Strings containing both the words 'grotto' and 'raven'
print("c. Contains both 'grotto' and 'raven':")
pattern_c = re.compile(r'(?=.*\bgrotto\b)(?=.*\braven\b)', re.IGNORECASE)

for line in lines:
    if pattern_c.search(line):
        print(f"Matched: {line.strip()}")

print("=" * 60)

# d. Capture the first word of an English sentence
print("d. First Word of Each Sentence:")
pattern_d = re.compile(r'(^|[.!?]\s+)([A-Za-z]+)')

for line in lines:
    matches = pattern_d.finditer(line)
    for match in matches:
        print(f"Sentence: {line.strip()}")
        print(f"First Word: {match.group(2)}")

print("=" * 60)
```

# Question 2

Implement an ELIZA-like program, using substitutions. Choose a domain in which
your program can legitimately engage in simple conversations. The program
should exit when the user types “BYE BYE”

```python
import re
import random

# Patterns and responses
rules = [
    (r".*\b(hi|hello|hey)\b.*",
     ["Hello! How can I help you with your studies?",
      "Hi there! What academic problem are you facing?"]),

    (r".*\bmy name is (\w+).*",
     ["Nice to meet you, \\1. How can I assist you today?"]),

    (r".*\b(i am|i'm) stressed\b.*",
     ["Why are you feeling stressed?",
      "Exams can be stressful. Have you prepared a study plan?"]),

    (r".*\b(i have|i've) exam\b.*",
     ["Which subject is your exam for?",
      "How are you preparing for your exam?"]),

    (r".*\b(i need help with) (.+)",
     ["What difficulty are you facing with \\2?",
      "Let's discuss \\2. Tell me more."]),

    (r".*\b(thank you|thanks)\b.*",
     ["You're welcome!",
      "Glad I could help."]),

    (r".*",
     ["Can you tell me more?",
      "Please explain further.",
      "I understand. Continue.",
      "That's interesting."])
]

print("===================================")
print(" ELIZA - Student Academic Counselor")
print(" Type 'BYE BYE' to exit.")
print("===================================")

while True:
    user = input("You: ")

    if user.upper() == "BYE BYE":
        print("ELIZA: Goodbye! Best of luck with your studies.")
        break

    responded = False

    for pattern, responses in rules:
        match = re.match(pattern, user, re.IGNORECASE)
        if match:
            response = random.choice(responses)

            # Replace captured groups in the response
            for i in range(1, len(match.groups()) + 1):
                response = response.replace(f"\\{i}", match.group(i))

            print("ELIZA:", response)
            responded = True
            break

    if not responded:
        print("ELIZA: Could you explain that differently?")
```

# Question 3
Implement a simple rule-based Text tokenizer for the English language using
regular expressions. Your tokenizer should consider punctuation and special
symbols as separate tokens. Contractions like "isn't" should be treated as 2 tokens: "is" and "n't". Also identify abbreviations (eg, U.S.A) and internal hyphenation (eg., ice-cream) as single tokens.

```python
import re

def tokenize(text):
    # Split contractions first
    text = re.sub(r"\b(\w+)n't\b", r"\1 n't", text)
    text = re.sub(r"\b(\w+)'(re|ve|ll|d|m|s)\b", r"\1 '\2", text)

    # Regular expression for tokens
    pattern = r"""
        (?:[A-Za-z]\.){2,}[A-Za-z]?\.?      # Abbreviations (U.S.A.)
      | [A-Za-z]+(?:-[A-Za-z]+)+           # Hyphenated words (ice-cream)
      | n't                                # n't contraction
      | '(?:re|ve|ll|d|m|s)                # 're, 've, 'll, etc.
      | [A-Za-z]+                          # Words
      | \d+(?:\.\d+)?                      # Numbers
      | [.,!?;:()"[\]{}]                   # Punctuation
    """

    return re.findall(pattern, text, re.VERBOSE)


# Read input from file
with open("input_3.txt", "r", encoding="utf-8") as f:
    text = f.read()

tokens = tokenize(text)

print("Input Text:\n")
print(text)

print("\nTokens:\n")
for token in tokens:
    print(token)
```

# Question 4

Design and implement a Finite State Automata(FSA) that accepts English plural
nouns ending with the character ‘y’, e.g. boys, toys, ponies, skies, and puppies but not boies or toies or ponys. (Hint: Words that end with a vowel followed by ‘y’ are appended with ‘s' and will not be replaced with “ies” in their plural form).

```python
def plural_fsa(word):
    vowels = "aeiouAEIOU"

    # Check words ending with "ies"
    if word.endswith("ies"):
        root = word[:-3]
        if len(root) > 0 and root[-1] not in vowels:
            return True
        return False

    # Check words ending with "ys"
    elif word.endswith("ys"):
        root = word[:-2]
        if len(root) > 0 and root[-1] in vowels:
            return True
        return False

    else:
        return False


# Read words from file
with open("input_4.txt", "r") as file:
    words = file.read().split()

print("Word\t\tResult")
print("-" * 25)

for word in words:
    if plural_fsa(word):
        print(f"{word}\t\tAccepted")
    else:
        print(f"{word}\t\tRejected")
```

# Question 5

Design and implement a Finite State Transducer(FST) that accepts lexical forms of English words(e.g. shown below) and generates its corresponding plurals, based on the e-insertion spelling rule є => e / {x,s,z}^ __ s#
^ is the morpheme boundary and # - word boundary

Input      --   Output

fox^s#     --   foxes

boy^s#     --   boys

```python
def plural_fst(word):
    # Remove morpheme (^) and word (#) boundaries
    if "^s#" not in word:
        return "Invalid lexical form"

    root = word.replace("^s#", "")

    # Apply e-insertion rule
    if root[-1] in ['x', 's', 'z']:
        return root + "es"
    else:
        return root + "s"


# Read lexical forms from file
with open("input_5.txt", "r") as file:
    words = file.read().split()

print("Lexical Form\tOutput")
print("-" * 35)

for word in words:
    print(f"{word}\t\t{plural_fst(word)}")
```

# Question 6

Implement a Byte Pair tokenizer for English using a small, representative corpus. Print each intermediate step in vocabulary creation. Discuss your choices and their impact.

```python
from collections import Counter

# Small English corpus
corpus = [
    "low",
    "lowest",
    "newer",
    "wider"
]

# Convert each word into characters + end-of-word symbol
vocab = Counter()

for word in corpus:
    chars = " ".join(list(word)) + " </w>"
    vocab[chars] += 1


def get_stats(vocab):
    """Count frequency of adjacent symbol pairs."""
    pairs = Counter()

    for word, freq in vocab.items():
        symbols = word.split()

        for i in range(len(symbols) - 1):
            pairs[(symbols[i], symbols[i + 1])] += freq

    return pairs


def merge_vocab(pair, vocab):
    """Merge the most frequent pair."""
    new_vocab = {}

    bigram = " ".join(pair)
    replacement = "".join(pair)

    for word in vocab:
        new_word = word.replace(bigram, replacement)
        new_vocab[new_word] = vocab[word]

    return new_vocab


print("Initial Vocabulary\n")

for word in vocab:
    print(word)

num_merges = 10

for i in range(num_merges):

    pairs = get_stats(vocab)

    if not pairs:
        break

    best = max(pairs, key=pairs.get)

    print("\n----------------------------------")
    print("Iteration", i + 1)
    print("Most frequent pair:", best)
    print("----------------------------------")

    vocab = merge_vocab(best, vocab)

    print("Vocabulary after merge:\n")

    for word in vocab:
        print(word)
```
